```python
# Task 3 :- Performing EDA on Students Performance Dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Cleaned_StudentsPerformance.csv")

# Show Statistics
print(df.describe())
```

```python
# Average Scores
print("\nMaths Average Score : ", df['math_score'].mean())
print("\nReading Average Score : ", df['reading_score'].mean())
print("\nWriting Average Score : ", df['writing_score'].mean())
```

```python
# Gender Count
print("\nGender Distribution : ")
print(df['gender'].value_counts())
```

```python
# Maths Score Distribution
plt.hist(df['math_score'], bins=20)
plt.title("Maths Score Distribution")
plt.xlabel("Maths score")
plt.ylabel("Counts")
plt.show()
```

```python
# Gender vs Maths Distribution
sns.boxplot(x='gender', y='math_score', data=df)
plt.title("Gender vs Maths Distribution")
plt.show()
```

```python
# Reading vs Writing Score
sns.scatterplot(x='reading_score', y='writing_score', data=df)
plt.title("Reading Score vs Writing Score")
plt.show()
```
